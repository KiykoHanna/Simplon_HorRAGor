# Nœud de Protection (Guardrail) — Защитный узел
---

## Основные понятия

- Детерминированный перехват запросов (interception déterministe);
- Защита входных данных (sécurisation des entrées);
- Обнаружение Prompt Injection (détection des injections de prompts);
- Проверка формата данных (validation de format).

Цель: Превратить входную точку графа в защитный барьер, который способен остановить вредоносные или некорректные запросы до того, как они попадут в основной LLM.

Если Evaluator (судья) проверяет качество работы в конце, то Guardrail действует в самом начале. До этого момента агент принимал любой текст пользователя. Это опасно, потому что пользователь может:

- отправить несуществующий ID;
- забыть обязательный параметр;
- попытаться взломать промпт (Prompt Injection);
- передать данные неверного формата.

Теперь каждый запрос сначала попадает в Guardrail. Он принимает одно из двух решений:

- Запрос безопасен, Если:

        - идентификатор имеет правильный формат;
        - необходимые данные существуют;
        - вредоносный текст не обнаружен.

        Тогда запрос передается основному агенту.

- Запрос опасен, Если:

        - обнаружена Prompt Injection;
        - отсутствует обязательный идентификатор;
        - указан несуществующий объект;
        - формат неверен.

        Тогда запрос сразу отклоняется. Основной LLM даже не запускается.
