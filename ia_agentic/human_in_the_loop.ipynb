# Human-in-the-Loop (HITL) — Человек в цикле принятия решений
---

- Модератор (Le modérateur)
- Прерывание графа (interrupt_before)
- Точки останова (Breakpoints)
- Сохранение состояния (MemorySaver / Checkpointer)
- Приостановленное состояние (Suspended State)

Цель

Обеспечить безопасность выполнения критически важных действий, заставляя граф останавливаться перед выполнением опасной операции и ждать подтверждения человека.

Если Tools дают агенту руки, то механизм Human-in-the-Loop дает ему ремень безопасности. До этого момента агент работал полностью автономно, Но в реальных приложениях существуют действия, которые нельзя доверять только LLM.

- подтверждение банковского перевода;
- удаление базы данных (DROP TABLE);
- удаление файлов;
- отправка писем клиентам;
- изменение медицинской информации.

Когда LLM собирается вызвать Tool,граф:

- сохраняет свое текущее состояние;
- полностью останавливается;
- ожидает решения человека.

После этого человек может:

- проверить, какой Tool будет вызван;
- разрешить выполнение;
- запретить выполнение.

### Практическая реализация

Кроме графа появляется MemorySaver. Именно он запоминает состояние графа.

```python
memory = MemorySaver()

app = workflow.compile(
    checkpointer=memory,
    interrupt_before=["tools"]
)
```
