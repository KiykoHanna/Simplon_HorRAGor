```python
import polars as pl

df = pl.read_parquet("../data/gold/gold_movies.parquet")
df = df.group_by(df.columns).first()
print(print(df.head(10)))
```

```python
dups = (
    df.group_by("master_id")
      .len()
      .filter(pl.col("len") > 1)
      .select("master_id")
)

df.filter(pl.col("master_id").is_in(dups["master_id"]))
```
